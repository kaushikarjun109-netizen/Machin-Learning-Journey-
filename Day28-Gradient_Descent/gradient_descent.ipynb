```python
import numpy as np
import pandas as pd 
```

```python
from sklearn.datasets import make_regression
```

```python
x,y = make_regression(n_samples=4 , n_features=1 , n_informative=1 , n_targets=1 , noise= 80 , random_state=13)
```

```python
import matplotlib.pyplot as plt 
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='orange')
```

```python
# Lets apply gradient descent assuming the slope value m = 78.35
#  and let's assume the starting value of intercept b = 0
y_pred = ((78.35*x)+0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='orange',label='OLS')
plt.plot(x,y_pred,color='green',label='b=0')
plt.legend()
plt.show()
```

```python
m = 78.35
b=0
loss_slope = -2*np.sum(y-m*x.ravel()-b)
loss_slope
```

```python
# Lets taking learning rate lr=0.1
lr=0.1
step_size =lr*loss_slope
step_size
```

```python
b = b- step_size
b
```

```python
y_pred1 = ((78.35*x)+b).reshape(4)
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='orange',label='OLS')
plt.plot(x,y_pred,color='green',label='b=0')
plt.plot(x,y_pred1,color='black',label='b={}'.format(b))
plt.legend()
plt.show()
```
