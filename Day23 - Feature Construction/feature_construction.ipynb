```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
```

```python
df = pd.read_csv('train.csv' , usecols=['Age' , 'Pclass' , 'SibSp' , 'Parch' , 'Survived'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
x = df.iloc[:,1:]
y = df.iloc[: ,0]
```

```python
x.head()
```

```python
y.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring='accuracy' , cv=20))
```

# Applying Feature Construction

```python
x['Family_size'] = x['SibSp'] + x['Parch'] + 1
```

```python
x.head()
```

```python
def func(num):
    if num == 1:
        return 0  # alone
    elif num >1 and num <=4:
        return 1 # small family
    else:
        return 2 # large family
    
```

```python
func(3)
```

```python
func(5)
```

```python
func(1)
```

```python
x['Family_type'] = x['Family_size'].apply(func)
```

```python
x.head()
```

```python
x.drop(columns=['SibSp' , 'Family_size' , 'Parch'] , inplace=True)
```

```python
x.head()
```

```python
np.mean(cross_val_score(LogisticRegression() , x , y , scoring='accuracy' , cv =20))
```
