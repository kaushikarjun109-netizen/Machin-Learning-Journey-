```python
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
df['number'].unique()
```

```python
fig = df['number'].value_counts().plot.bar()
fig.set_title('Passenger Travelling with')
```

```python
#extracting numbers
df['number_numerical'] = pd.to_numeric(df['number'] , errors ='coerce' , downcast='integer')
```

```python
# extracting categorical data
df['number_categorical'] = np.where(df['number_numerical'].isnull(),df['number'] , np.nan)
```

```python
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['number_numerical'] = df['number_numerical'].where(
    df['number_numerical'].notna(), 0 
)
```

```python
df['number_numerical']
```

```python
df['Cabin_num'] = df['Cabin'].str.extract('(\d+)') #Capture numerical values 
df['Cabin_cat'] = df['Cabin'].str[0] #Capture the first letter 
```

```python
df.head()
```

```python
#Extracting the last bit of ticket as number 
df['ticket_num'] = df['Ticket'].apply(lambda s : s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                 errors='coerce',
                                 downcast='integer')

# Extract the first letter of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s : s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit() , np.nan ,  df['ticket_cat'])

df.head()
```
