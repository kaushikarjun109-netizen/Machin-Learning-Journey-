```python
import numpy as np 
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

```python
df.drop(columns=['PassengerId' , 'Name' , 'Ticket' , 'Cabin'] , inplace=True)
```

```python
df.head()
```

# Train Test Split

```python
x_train , x_test , y_train , y_test = train_test_split(df.drop(columns=['Survived']) , df['Survived'] , test_size=0.2 , random_state=42)
```

```python
x_train.head(4)
```

```python
y_train.head(4)
```

```python
df.isnull().sum()
```

```python
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent') # the most frequently city come will be take a place of missing values 
#train one
x_train_age = si_age.fit_transform(x_train[['Age']])
x_train_embarked = si_embarked.fit_transform(x_train[['Embarked']])
#test one 
x_test_age = si_age.fit_transform(x_test[['Age']])
x_test_embarked = si_embarked.transform(x_test[['Embarked']])
```

```python
x_train_age , x_test_age , x_test_embarked , x_train_embarked
```

# Encoding

```python
ohe_embarked = OneHotEncoder(sparse_output=False , handle_unknown='ignore')
ohe_sex = OneHotEncoder(sparse_output=False , handle_unknown='ignore')
x_train_embarked = ohe_embarked.fit_transform(x_train[['Embarked']])
x_train_sex = ohe_sex.fit_transform(x_train[['Sex']])
#test one 
x_test_embarked = ohe_embarked.transform(x_test[['Embarked']])
x_test_sex = ohe_sex.transform(x_test[['Sex']])
```

```python
x_train_rem = x_train.drop(columns=['Age' , 'Sex' , 'Embarked'])
x_test_rem = x_test.drop(columns=['Age' , 'Sex' , 'Embarked'])
```

```python
x_train_transformed = np.concatenate((x_train_rem , x_train_age , x_train_sex , x_train_embarked),axis=1)
x_test_transformed = np.concatenate((x_test_rem , x_test_age , x_test_sex , x_test_embarked),axis=1)
```

```python
x_train_transformed.shape
```

```python
clf = DecisionTreeClassifier()
clf.fit(x_train_transformed , y_train)
```

```python
y_predi = clf.predict(x_test_transformed)
y_predi
```

```python
print("y_test:", len(y_test))
print("y_train:", len(y_train))
print("x_test_transformed:", x_test_transformed.shape)
print("x_train_transformed:", x_train_transformed.shape)
```

```python
y_predict = clf.predict(x_test_transformed)
print("y_predict:", len(y_predict))
    
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_predi)
```
