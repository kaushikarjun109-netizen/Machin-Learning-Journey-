```python
import numpy as np
import pandas as pd 
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
x,y = make_regression(n_samples=100,n_features=1,n_targets=1,n_informative=1,noise=20,random_state=13)
```

```python
x_train , x_test, y_train , y_test = train_test_split(x,y,test_size=0.2)
```

```python
class MeraLasso:
    def __init__(self,alpha = 0.1):
        self.intercept = None
        self.coef = None
        self.alpha = alpha
    def fit(self,x_train,y_train):
        num=0
        den=0
        for i in range(x_train.shape[0]):
            num = num + (y_train[i] - y_train.mean())*(x_train[i] - x_train.mean())
            den = den + (x_train[i] - x_train.mean())*(x_train[i] - x_train.mean())
        if self.alpha<num:
            self.coef = (num-self.alpha)/den
            self.intercept = y_train.mean() - self.coef*x_train.mean()
        elif -self.alpha>num:
            self.coef = (num+self.alpha)/den
            self.intercept = y_train.mean() - self.coef*x_train.mean()
        else:
            self.coef = 0
            self.intercept = y_train.mean() - self.coef*x_train.mean()
        print(self.coef,self.intercept)
    def predict(self ,x_test):
        return self.coef*x_test + self.intercept
```

```python
lasso = MeraLasso(alpha=1000)
lasso.fit(x_train,y_train)
```
