```python
import numpy as np
import pandas as pd
df = pd.read_csv('covid_toy.csv')
df.head()
```

# Train Test split

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(df.drop('has_covid' , axis=1) , df['has_covid'] , test_size= 0.2 , random_state=0)
x_train.shape , x_test.shape
```

```python
df['fever'].isnull().sum # so fever column have 10 missing values
```

# Aam Zindagi 

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
# adding simple imputer to fever col
si = SimpleImputer()
x_train_fever = si.fit_transform(x_train[['fever']]) # it will replace all missing values by it's mean value of column.
# also test data
x_test_fever = si.fit_transform(x_test[['fever']])
x_train_fever.shape
```

```python
#Ordinal Encoding -> cough
oe = OrdinalEncoder(categories=[['Mild' , 'Strong']])
x_train_cough = oe.fit_transform(x_train[['cough']])
#also test one
x_test_cough = oe.fit_transform(x_test[['cough']])
x_train_cough.shape
```

```python
#One Hot Encodinf -> gender , city
ohe = OneHotEncoder(drop='first')
x_train_gender_city = ohe.fit_transform(x_train[['gender' , 'city']]).toarray()
#one test data also
x_test_gender_city = ohe.fit_transform(x_test[['gender' , 'city']]).toarray()
x_test_gender_city.shape , x_train_gender_city.shape
```

```python
#Extracting age
x_train_age = x_train.drop(columns=['gender' ,'fever', 'cough' , 'city'])
x_test_age = x_test.drop(columns=['gender' ,'fever', 'cough' , 'city'])
x_train_age
```

```python
# conatenate everone 
x_train_transformed = np.concatenate((x_train_age , x_train_fever , x_train_gender_city, x_train_cough ), axis=1)
#test one also 
x_test_transformed = np.concatenate((x_test_age , x_test_fever , x_test_gender_city, x_test_cough ), axis=1)
x_train_transformed.shape
```

# Mentos Zindagi by Column transform

```python
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(transformers=[
    ('tnf1' , SimpleImputer() , ['fever']),
    ('tnf2' , OrdinalEncoder(categories=[['Mild' ,'Strong']]) , ['cough']),
    ('tnf3', OneHotEncoder(sparse_output=False ,drop='first') , ['gender' , 'city'])
] , remainder='passthrough')
```
